```python
#Libraries, loops, and lists
#import two standard libraries

import numpy as np
import pandas as pd
```

```python
#initialize an array
#make an array of numbers 0 to 100 (exclusive)
random_list = np.arange(stop = 100)
#consult documentation for further function paramteres/outputs, etc.
```

```python
#print is built-in in python
print(random_list)
```

```python
#A1 use np.arange to create a new list
#called even_list that contains all the even numbers from 0 to 100 (exclusive)

even_list_1 = np.arange(0, 100, 2)
print(even_list_1)
```

```python
#the [] allows us to access elements from the array
print(random_list[0])

#use the : operator to access elements across a certain range
print(random_list[5 : 15])
#note that the last element is exclusive

#we also ust access a number at either end of our list using : to indicate we want python to go all the way to the end
print(random_list[95 : ])
print(random_list[:5])
```

```python
#A2

#print the element at the 90th index of the array
print(random_list[90])

#print the element from index 10 to index 21
print(random_list[10:22])

#print the last 10 elements of the array
print(random_list[(len(random_list) - 10) :])
```

```python
#doing mathematical operations on our list
#want the list to start at 1 instead of 0
#add 1 to each element in the array
random_list = random_list + 1
print(random_list)
```

```python
#do the same with natural nlog, using np.log command
#take the natural log of each element in the array, and printing the result

print(np.log(random_list))
```

```python
#boolean indexing can also be performed element-wise across an array
#boolean index array that shows where all the odds(falses) are:
print(random_list % 2 == 0)

#taking in the boolean index array into the new array allows us to filter out the falses
even_list_2 = random_list[random_list % 2 == 0]
print(even_list_2)
```

```python
#initialize an empty array
even_list_3 = np.array([])

#for i in range just means for each value (i) in the range from 0 to 101 (exclusive)
for i in range(0,101):
    #we check if the value (i) is even (when divided by 2, remainder = 0)
    if i% 2 == 0:
        #use the np.append function
        even_list_3 = np.append(arr = even_list_3, values = i)
print(even_list_3)
```

```python
#make an array of numbers 0 to 200 (exclusive)
big_list = np.arange(200)

#multiply numbers resuling in a list
even_list_4 = [2*x for x in big_list if x<=50] 

#convert to an array
even_list_4 = np.array(even_list_4)

print(even_list_4)
```
